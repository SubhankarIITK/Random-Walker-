## 2. Statistical Properties

One of the key theoretical results is that the **Mean Squared Displacement (MSD)** is equal to the number of steps $n$, regardless of the dimension $d$.
$$ E[|X_n|^2] = n $$

```python
from src.simulation.ensemble import ensemble_final_positions
from src.analysis.statistics import compute_ensemble_statistics

n_walkers = 2000
n_steps = 1000

print(f"Simulating ensemble of {n_walkers} walkers for {n_steps} steps in {d}D...")
final_pos = ensemble_final_positions(d, n_steps, n_walkers, seed=123)

stats = compute_ensemble_statistics(final_pos, n_steps, d)
print(f"Theoretical MSD: {stats['msd_theory']:.2f}")
print(f"Simulated MSD:   {stats['msd_sim']:.2f} ± {stats['msd_se']:.2f}")
print(f"Relative Error:  {stats['pct_error_msd']:.2f}%")
```

## 3. The Recurrence-Transience Transition
George Pólya proved in 1921 that a random walk on a $d$-dimensional grid is *recurrent* (returns to the origin with probability 1) for $d = 1, 2$, but is *transient* (probability of return is < 1) for $d \geq 3$.

```python
from src.theory.recurrence import recurrence_classification, theoretical_return_probability

for dim in range(1, 6):
    prob = theoretical_return_probability(dim)
    cls = recurrence_classification(dim)
    print(f"{dim}D: {cls} (Return Probability = {prob:.4f})")
```
